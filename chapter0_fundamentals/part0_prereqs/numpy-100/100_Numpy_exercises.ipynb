#### 63. Create an array class that has a name attribute (★★☆)

````python
class special(np.ndarray):
    def __new__(cls, input, name="default"):
        obj = np.asarray(input).view(cls)
        obj.name = name
        return obj

    def __array_finalize__(self, obj):
        if obj is None:
            return
        self.name = getattr(obj, "name", "default")
````

#### 80. Consider an arbitrary array, write a function that extracts a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)

````python
a = np.random.rand(5,5,5)

def e(a, c, s, f=0):
    c = np.array(c)
    s = np.array(s)

    out = np.full(s, f, dtype=a.dtype)

    half = s // 2
    start = c - half
    end = start + s

    s_start = np.maximum(start, 0)
    s_end = np.minimum(end, np.array(a.shape))

    d_start = s_start - start
    d_end = d_start + (s_end - s_start)

    s_slices = tuple(slice(s, e) for s, e in zip(s_start, s_end))
    d_slices = tuple(slice(s, e) for s, e in zip(d_start, d_end))

    out[d_slices] = a[s_slices]

    return out

e(a, c = (0,0,0), s = (3,3,3), f=-1)
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?

````python
x = np.random.randint(0,6, size =1)
y = np.random.randint(0,6, size = 1)

d = np.sqrt(np.diff(x)**2 + np.diff(y)**2)

s = np.concatenate(([0], np.cumsum(d)))

s_n = np.linspace(0, s[-1], 20)

x_n = np.interp(s_n, s, x)
y_n = np.interp(s_n, s, y)

print(x_n)
print(y_n)
````
